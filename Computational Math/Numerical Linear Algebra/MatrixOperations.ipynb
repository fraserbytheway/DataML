### Matrix and Vector Operations

#### Basics

```python
import numpy as np


A = np.array([[2, 0, -1], [5, 7, 3], [-4, 1, 8], [9, 4, 6]])
B = np.array([[1, 1, 2], [2, -1, 6], [9, 5, 5]])
x = np.array([3, 0, -5, 2])


print("AB\n", A @ B, "\n")
print("A.Tx\n", A.T @ x, "\n")
print("BA.T\n", B @ A.T, "\n")
print("A.TA + B\n", A.T@A + B, "\n")
print("x.Tx - 3\n", x.T @ x - 3, "\n")
```

#### BLAS Routine

```python
def gemm(alpha, A, B, beta, C):

    m, n = C.shape # A a m * p matrix, B a p * n matrix
    _, p = A.shape

    for i in range(m):
        for j in range(n):
            c = 0
            for k in range(p):
                c += A[i][k] * B[k][j]
            C[i][j] = alpha * c + beta * C[i][j]

    return C

n = 2^10
A = np.random.random((n, n))
B = np.random.random((n, n))
C = np.random.random((n, n))
alpha = 3
beta = -2


gemm(alpha, A, B, beta, C)
```
