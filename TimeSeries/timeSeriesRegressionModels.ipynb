### Time Series Regression Models

Forecast time series of interest y, assuming it has a linear relationship with other time series x.
* forecast mon thly sales y using total advertising spend x as a predictor
* forecast dailty electricity demand y using temperature x1 and day of week x2 as predictors

**Assumptions** - linear regression error term
* have mean 0, otherwise forecast systematically biased
* not autocorrelated; otherwise forecsts will be inefficient
* Unrelated to the predictor variables

Also that the model is a reasonable approximation to reality. Relationship between the forecast variable and the predictor variables satisfies a linear equation


#### Evaluating the Regression Model

After fitting a regression model, necessary to plot the residuals to check that the assumptions of the model have been satsified.
1. ACF plot of residuals. Check for autocorrelation
2. Lagrange Multiplier test for serial correlation
3. QQ plot
4. Residual plot against predictors (want no correlation here)
5. Residuals against fitted values
6. Outliers and influential observations

#### Useful predictors

* Trend. Simply use t as a predictor
* Dummy Variables (public holiday, outliers)
* Seasonal Dummy Variables
* Intervention Variables - model interventions that may have affected the variable to be forecast. For example advertising expenditure, industrial action. Use a spike variable, which is a dummy variable that takes value only in period of the intervention
* Trading days - number of business days in the month
* Distributed lag (advertising expenditure can have a delayed response)
* Public Holidays
* Fourier series for seasonality effects

#### Selecting Predictors

Common approach is to do a multiple linear regression on all the predictors, and disregard all variables whose p-values are greater than 0.05. Statistical significance doesn't always indicate the predictive value, and p-values can be misleading when two or more predictors are correlated. Instead use a measure of predictive accuracy. Five such measures are cv, AIC, BIC, adj R2, AICc.

* Adj R2 penalises more predictors
* Cross Validation for finding overfitting
* AIC criterion

When possible, all possible regressions should be fitted. Best model should be selected based on measures above. An alternative to this is stepwise regression, which can be a better approach for large number of predictors.


#### Forecasting with Regression

Ex-ante forecasts - made using only information available in advance
Ex-post forecasts - made using later information on the predictors
Scenario based forecasts - analysis of predictions under different scenarios


Can be very difficult to obtain the right input predictor set for ex-ante forecasts. Can instead do a h-step ahead forecast, where prediction h steps into the future, made from observations up to current time.

#### Nonlinear Regression

many cases in which a non-linear functional form is more suitable.
* log-log function
* piecewise linear transform (regression splines)
* non linear trend, use t^2

#### Correlation, Causation and Forecasting

confounder - variable that influences both the response and at least one predictor variable.
Correlated predictors - when two or more predictors are highly correlated, can be challenging to accurately seperate their individual effcts
Multicolinearity - two predictors highly correlated, they both provide similar information. With high correlation, estimation of the regression coefficients is computationally difficult. Uncertainty associated with individual regression coefficients will be large. Forecasts will be unreliable if the value of future predictors are outside the range of the historical value of the predictors. Problem espacially present when multicollinearity is present.
